BUBBLE SORT

```python
def bubble_sort(arr):
  n = len(arr)
  for i in range(n-1):
    for j in range(n - i - 1):
      if arr[j] > arr[j + 1]:
        arr[j], arr[j + 1] = arr[j + 1], arr[j]
  return arr

# Makenya gini nih
data = [64, 34, 25, 12, 22, 11, 90]
sorted_data = bubble_sort(data.copy())
print("Data sebelum di urutin:", data)
print("Data setelah Bubble Sort: ", sorted_data)
```

SELECTION SORT

```python
def selection_sort(arr):
  n = len(arr)
  for i in range(n - 1):
    min_index = i
    for j in range(i + 1, n):
      if arr[j] < arr[min_index]:
        min_index = j
    arr[i], arr[min_index] = arr[min_index], arr[i]
  return arr

  # makenya gini nih
data = [64, 25, 12, 22, 11]
sorted_data = selection_sort(data.copy())
print("Data sebelum sorting:", data)
print("Data setelah Selection Sort:", sorted_data)
```

INSERTION SORT

```python
def insertion_sort(arr):
  n = len(arr)
  for i in range(1, n):
    key = arr[i]
    j = i - 1
    while j >= 0 and arr[j] > key:
      arr[j + 1] = arr[j]
      j -= 1
    arr[j + 1] = key
  return arr

  # makenya gini nih
data = [64, 34, 25, 12, 22, 11, 90]
sorted_data = insertion_sort(data.copy())
print("Data sebelum diurutkan:", data)
print("Data setelah Insertion Sort:", sorted_data)
```

MERGE SORT

```python
def merge_sort(arr):
  if len(arr) > 1:
    mid = len(arr) // 2
    left_half = arr[:mid]
    right_half = arr[mid:]

    merge_sort(left_half)
    merge_sort(right_half)

    i = j = k = 0

    while i < len(left_half) and j < len(right_half):
      if left_half[i] < right_half[j]:
        arr[k] = left_half[i]
        i += 1
      else:
        arr[k] = right_half[j]
        j += 1
        k +=1
    while i < len(left_half):
      arr[k] = left_half[i]
      i += 1
      k += 1
    while j < len(right_half):
      arr[k] = right_half[j]
      j += 1
      k += 1
  return arr

  # makenya gini nih
data = [64, 25, 12, 22, 11]
sorted_data = merge_sort(data.copy())
print("Data sebelum sorting:", data)
print("Data setelah Merge Sort:", sorted_data)
```

QUICK SORT

```python
def quick_sort(arr):
  if len(arr) <= 1:
    return arr

  pivot = arr[len(arr) // 2]
  left = [x for x in arr if x < pivot]
  middle = [x for x in arr if x == pivot]
  right = [x for x in arr if x > pivot]

  return quick_sort(left) + middle + quick_sort(right)

  # makenya gini nihhhh
data = [64, 25, 12, 22, 11]
sorted_data = quick_sort(data.copy())
print("Data sebelum sorting:", data)
print("Data setelah Quick Sort:", sorted_data)
```

Sistem Manajemen Data Pengguna (Rombel 01)

Sebuah aplikasi e-commerce memiliki daftar pelanggan yang harus diurutkan
berdasarkan nama (A-Z), umur, atau total belanja. Implementasikan algoritma
sorting untuk mengatur data pelanggan

```python
def merge_sort_customers(customers, key='total_belanja', descending=True):
    """
    Mengurutkan list pelanggan berdasarkan key tertentu menggunakan Merge Sort.

    Parameters:
    - customers: list of dict (data pelanggan)
    - key: nama field yang dijadikan acuan sorting
    - descending: True untuk urutan terbesar ke terkecil, False untuk sebaliknya
    """
    if len(customers) <= 1:
        return customers.copy()

    mid = len(customers) // 2
    left = merge_sort_customers(customers[:mid], key, descending)
    right = merge_sort_customers(customers[mid:], key, descending)

    return merge_customers(left, right, key, descending)

def merge_customers(left, right, key, descending):
    """Fungsi bantuan untuk menggabungkan dua list pelanggan yang terurut."""
    result = []
    i = j = 0

    while i < len(left) and j < len(right):
        if descending:
            # Urutan descending: nilai lebih besar didahulukan
            if left[i][key] >= right[j][key]:
                result.append(left[i])
                i += 1
            else:
                result.append(right[j])
                j += 1
        else:
            # Urutan ascending: nilai lebih kecil didahulukan
            if left[i][key] <= right[j][key]:
                result.append(left[i])
                i += 1
            else:
                result.append(right[j])
                j += 1

    result.extend(left[i:])
    result.extend(right[j:])
    return result

# Data pelanggan contoh
customers = [
    {"nama": "Budi Santoso", "umur": 25, "total_belanja": 1500000},
    {"nama": "Ani Wijaya", "umur": 30, "total_belanja": 2500000},
    {"nama": "Citra Dewi", "umur": 22, "total_belanja": 900000},
    {"nama": "Dedi Firmansyah", "umur": 35, "total_belanja": 3200000},
    {"nama": "Eka Lestari", "umur": 28, "total_belanja": 2100000},
]

print("="*60)
print("STUDI KASUS 1: Sistem Manajemen Data Pengguna E-commerce")
print("="*60)
print("\nData sebelum sorting:")
for cust in customers:
    print(f"  {cust['nama']:15} | Umur: {cust['umur']:2} | Total Belanja: Rp{cust['total_belanja']:>10,}")

# Urutkan berdasarkan total belanja dari tertinggi ke terendah
sorted_customers = merge_sort_customers(customers, key='total_belanja', descending=True)

print("\n📊 Data setelah diurutkan berdasarkan TOTAL BELANJA (Tertinggi → Terendah):")
print("-" * 60)
for i, cust in enumerate(sorted_customers, 1):
    print(f"{i:2}. {cust['nama']:15} | Umur: {cust['umur']:2} | Total Belanja: Rp{cust['total_belanja']:>10,}")

# Bonus: Urutkan berdasarkan nama (A-Z)
sorted_by_name = merge_sort_customers(customers, key='nama', descending=False)

print("\n📊 Bonus: Data setelah diurutkan berdasarkan NAMA (A-Z):")
print("-" * 60)
for i, cust in enumerate(sorted_by_name, 1):
    print(f"{i:2}. {cust['nama']:15} | Umur: {cust['umur']:2} | Total Belanja: Rp{cust['total_belanja']:>10,}")
```
