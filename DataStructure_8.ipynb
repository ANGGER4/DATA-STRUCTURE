REPRESENTASI GRAF DENGAN ADJACENCY LIST

```python
# Random graf menggunakan adjacency list
graph = {
'A': ['B', 'C'],
'B': ['A', 'D', 'E'],
'C': ['A', 'F'],
'D': ['B'],
'E': ['B', 'F'],
'F': ['C', 'E']
}

# Menampilkan graf
def print_graph(graf) :
  for node in graf:
    print (f" {node} -> {graf [node] }")

print_graph (graph)
```

TRAVERSAL GRAF MENGGUNAKAN BFS

```python
from collections import deque

def bfs(graph, start):
  visited = set()
  queue = deque([start])
  visited.add(start)
  while queue:
    vertex = queue.popleft()
    print(vertex, end=" ")

    # Explore neighbors
    for neighbor in graph[vertex]:
      if neighbor not in visited:
        visited.add(neighbor)
        queue.append(neighbor)

# Contoh pemanggilan
bfs(graph, 'A')
```

TRAVERSAL MENGGUNAKAN DFS

```python
def dfs(graph, start, visited=None):
  if visited is None:
    visited = set ()

  if start not in visited:
    print(start, end=" ")
    visited.add(start)
    for neighbour in graph[start]:
      dfs(graph, neighbour, visited)

# Contoh pemanggilan
dfs(graph, 'A')
```

LATIHAN PERSONAL

```python
# Lengkapi fungsi untuk menghitung jumlah tetangga dari tiap simpul dalam graf
def count_neighbors (graph):
  result = {}
  for node in graph:
    # Hitung jumlah tetangga
    result[node] = len(graph[node])
  return result

  # Memanggil fungsi count_neighbors dan menampilkan hasilnya
neighbor_counts = count_neighbors(graph)
print("Jumlah tetangga per simpul:")
for node, count in neighbor_counts.items():
  print(f"Node {node}: {count} tetangga")
```

Tugas 2

```python
from collections import deque

def bfs_return_list(graph, start):
    visited = set()
    queue = deque([start])
    visited.add(start)
    result = []

    while queue:
        vertex = queue.popleft()
        result.append(vertex)

        for neighbor in graph[vertex]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

    return result

# Contoh pemanggilan
urutan_bfs = bfs_return_list(graph_baru, 'A')
print(f"Urutan BFS: {urutan_bfs}")
```

Tugas 3

```python
def find_path(graph, start, end, path=None):
    if path is None:
        path = []

    path = path + [start]

    if start == end:
        return path

    if start not in graph:
        return None

    for neighbor in graph[start]:
        if neighbor not in path:
            new_path = find_path(graph, neighbor, end, path)
            if new_path:
                return new_path

    return None

# Contoh pemanggilan
jalur = find_path(graph_baru, 'A', 'F')
print(f"Jalur dari A ke F: {jalur}")

jalur_2 = find_path(graph_baru, 'A', 'D')
print(f"Jalur dari A ke D: {jalur_2}")
```

```python
def is_connected(graph):
    if not graph:
        return True

    # Mulai DFS dari simpul pertama
    start_node = list(graph.keys())[0]
    visited = set()

    def dfs_connected(node):
        visited.add(node)
        for neighbor in graph[node]:
            if neighbor not in visited:
                dfs_connected(neighbor)

    dfs_connected(start_node)

    # Cek apakah semua simpul telah dikunjungi
    return len(visited) == len(graph)

# Contoh pemanggilan
print(f"Apakah graf terhubung? {is_connected(graph_baru)}")

# Buat graf yang tidak terhubung untuk pengujian
graph_disconnected = {
    'A': ['B'],
    'B': ['A'],
    'C': ['D'],
    'D': ['C']
}
print(f"Apakah graf disconnected terhubung? {is_connected(graph_disconnected)}")
```
