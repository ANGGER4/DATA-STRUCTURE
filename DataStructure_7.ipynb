MANUAL BINARY TREE

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

# Membuat tree secara manual
root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.left.left = Node(4)
root.left.right = Node(5)

# Fungsi inorder traversal
def inorder(node):
    if node:
        inorder(node.left)
        print(node.data, end=' ')
        inorder(node.right)

print("Hasil inorder traversal dari binary tree:")
inorder(root)
```

BINARY TREE

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

class BinarySearchTree:
    def __init__(self):
        self.root = None

    def insert(self, root, data):
        if root is None:
            return Node(data)
        if data < root.data:
            root.left = self.insert(root.left, data)
        else:
            root.right = self.insert(root.right, data)
        return root

bst = BinarySearchTree()
bst_root = None # Change root to bst_root to avoid conflict with the manual tree's root
data_list = [50, 30, 70, 20, 40, 60, 80]
for value in data_list:
    bst_root = bst.insert(bst_root, value)
```

TRAVERSAL PADA BST

```python
def inorder(node):
    if node:
        inorder(node.left)
        print(node.data, end=' ')
        inorder(node.right)

def preorder(node):
    if node:
        print(node.data, end=' ')
        preorder(node.left)
        preorder(node.right)

def postorder(node):
    if node:
        postorder(node.left)
        postorder(node.right)
        print(node.data, end=' ')

print("Inorder Traversal:")
inorder(root)
print("\nPreorder Traversal:")
preorder(root)
print("\nPostorder Traversal:")
postorder(root)
```

Pencarian Nilai dalam BST

```python
def search(node, key):
    if node is None or node.data == key:
        return node
    if key < node.data:
        return search(node.left, key)
    return search(node.right, key)

# Uji pencarian pada BST (menggunakan bst_root)
key = 60
result = search(bst_root, key)
if result:
    print(f"{key} ditemukan dalam tree.")
else:
    print(f"{key} tidak ditemukan.")

key = 25
result = search(bst_root, key)
if result:
    print(f"{key} ditemukan dalam tree.")
else:
    print(f"{key} tidak ditemukan.")
```

Tugas 1: Program Struktur Binary Tree Manual

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

def inorder(node):
    if node:
        inorder(node.left)
        print(node.data, end=' ')
        inorder(node.right)

full_name = "Angger Naufal Abror"
NPM = "2505060020"
birth_date_dd = 9
birth_month_mm = 9

# Node root adalah jumlah huruf pada nama lengkap Anda.
root_value = len(full_name.replace(" ", "")) # Menghitung huruf, mengabaikan spasi

# Anak kiri root: jumlah huruf vokal (a, i, u, e, o) pada nama Anda.
vowels = "aeiouAEIOU"
vowel_count = sum(1 for char in full_name if char in vowels)

# Anak kanan root: dua digit terakhir dari NPM Anda.
NPM_last_two_digits = int(NPM[-2:])

# Membuat node-node
root = Node(root_value)
root.left = Node(vowel_count)
root.right = Node(NPM_last_two_digits)


# Anak kiri dari node kiri
if root.left:
    root.left.left = Node(birth_date_dd)

# Anak kanan dari node kanan
if root.right:
    root.right.right = Node(birth_month_mm)

# Menampilkan tree dengan inorder traversal.
print("Inorder Traversal dari Tree berdasarkan Identitas:")
inorder(root)
print()
```

Tugas 2: Program Binary Search Tree (BST)

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

class BinarySearchTree:
    def __init__(self):
        self.root = None

    # Metode untuk menyisipkan data ke dalam BST
    def insert(self, root, data):
        if root is None:
            return Node(data)
        if data < root.data:
            root.left = self.insert(root.left, data)
        else:
            root.right = self.insert(root.right, data)
        return root

full_name_bst = "Angger Naufal Abror"
NPM_bst = "2505060020"
birth_date_dd_bst = 9
birth_month_mm_bst = 9

# Tiga digit terakhir dari NPM
last_three_npm = int(NPM_bst[-3:])

# Tanggal lahir (hari dan bulan)
val_dd = birth_date_dd_bst
val_mm = birth_month_mm_bst

# ASCII dari huruf pertama dan kedua nama depan
first_name = full_name_bst.split()[0]
ascii_first_char = ord(first_name[0])
ascii_second_char = ord(first_name[1])

# Dua angka bebas
fav_num_1 = 9
fav_num_2 = 6

unique_numbers = {last_three_npm, val_dd, val_mm, ascii_first_char, ascii_second_char, fav_num_1, fav_num_2}
data_for_bst = list(unique_numbers)

print(f"Data numerik unik untuk BST: {data_for_bst}")


bst_instance = BinarySearchTree()
bst_root_tugas2 = None

for number in data_for_bst:
    bst_root_tugas2 = bst_instance.insert(bst_root_tugas2, number)

def inorder_bst(node):
    if node:
        inorder_bst(node.left)
        print(node.data, end=' ')
        inorder_bst(node.right)

def preorder_bst(node):
    if node:
        print(node.data, end=' ')
        preorder_bst(node.left)
        preorder_bst(node.right)

def postorder_bst(node):
    if node:
        postorder_bst(node.left)
        postorder_bst(node.right)
        print(node.data, end=' ')

print("\n--- Hasil Traversal BST ---")
print("Inorder Traversal:")
inorder_bst(bst_root_tugas2)
print("\nPreorder Traversal:")
preorder_bst(bst_root_tugas2)
print("\nPostorder Traversal:")
postorder_bst(bst_root_tugas2)
print()

# --- Bagian 6: Fungsi Pencarian dalam BST ---
def search_bst(node, key):
    if node is None or node.data == key:
        return node
    if key < node.data:
        return search_bst(node.left, key)
    return search_bst(node.right, key)

print("\n--- Hasil Pencarian dalam BST ---")


search_key_npm = int(NPM_bst[-2:])
result_npm = search_bst(bst_root_tugas2, search_key_npm)
if result_npm:
    print(f"Angka {search_key_npm} (dua digit terakhir NPM) ditemukan dalam BST.")
else:
    print(f"Angka {search_key_npm} (dua digit terakhir NPM) TIDAK ditemukan dalam BST.")


search_key_not_in_list = 99
result_not_in_list = search_bst(bst_root_tugas2, search_key_not_in_list)
if result_not_in_list:
    print(f"Angka {search_key_not_in_list} ditemukan dalam BST.")
else:
    print(f"Angka {search_key_not_in_list} TIDAK ditemukan dalam BST.")
```
